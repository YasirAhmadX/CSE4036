# Single layer perceptpron

```python
import numpy as np
import pandas as pd


def activation_func(value,actfn = "sigmoid"):    
    if actfn == "sigmoid":
        return (1/(1+np.exp(-value)))
    elif actfn == "tanh":
        return ((np.exp(value)-np.exp(-value))/(np.exp(value)+np.exp(-value)))
    else:
        return max(0,value) #relu

def perceptron_train(in_data,labels,alpha,actfn, bias = False):
    X=np.array(in_data)
    y=np.array(labels)
    weights=np.random.random(X.shape[1])
    
    #if bias:
        ##to be written
    
    original=weights
    
    for epoch in range(10):
        print(f"Epoch {epoch}/10")
        for key in range(X.shape[0]):
            z = np.matmul(np.transpose(weights),X[key])
            a=activation_func(z,actfn)
            yn=0
            if actfn == "sigmoid":    
                if a>=0.5:
                    yn=1
                else:
                    yn=0
            elif actfn == "tanh":
                if a>0:
                    yn=1
                else:
                    yn=0
            else:
                if a>5:
                    yn=1
                else:
                    yn=0
            weights=weights+alpha*(yn-y[key])*X[key]
            print('Iteration '+str(key)+': '+str(weights))
        print('Difference: '+str(weights-original))
    return weights

#  Testing and Score

def perceptron_test(in_data,label_shape,weights,actfn):
    X=np.array(in_data)
    y=np.zeros(label_shape)
    print(f"Testing:")
    for key in range(X.shape[1]):
        z = np.matmul(np.transpose(weights),X[key])
        a=activation_func(z,actfn)
        y[key]=0
        if actfn == "sigmoid":
            if a>=0.5:
                y[key]=1
            else:
                y[key]=0
        elif actfn == "tanh":
            if a>0:
                y[key] = 1
            else:
                y[key] = 0
        else:
            if a>0.3:
                yn=1
            else:
                yn=0
            #print(f"key:{key}\ty:{y}")
    return y
def score(result,labels):
    difference=result-np.array(labels)
    correct_ctr=0
    for elem in range(difference.shape[0]):
        if difference[elem]==0:
            correct_ctr+=1
    score=correct_ctr*100/difference.size
    print('Score='+str(score))
```

# Linear dataset

## Without bias sigmoid

```python
# Main code

# Dividing d_train into data and labels/targets
d_train_y = ANDdata['C']
d_train_X = ANDdata.drop(['C'], axis=1)

    # Dividing d_train into data and labels/targets
d_test_y = ANDdata['C']
d_test_X = ANDdata.drop(['C'], axis=1)

    # Learning rate
alpha = 0.55

    # Train
weights = perceptron_train(d_train_X, d_train_y, alpha, actfn="sigmoid")

    # Test
result_test = perceptron_test(d_test_X, d_test_y.shape, weights, actfn="sigmoid")

for i in range(len(result_test)):
    print(f"A:{d_test_X.iloc[i][0]}\tB:{d_test_X.iloc[i][1]}\tC:{d_test_y[i]}\tC_pred:{result_test[i]}")

from sklearn.metrics import log_loss
print(f"BCE {log_loss(result_test, d_test_y)}")

    # Calculate score
score(result_test, d_test_y)
```
